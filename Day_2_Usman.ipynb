# Game

```python
secret_number = 67

guess = int(input("Guess a number: "))

while guess != secret_number:
    if guess > secret_number:
        print("Little lower")
    else:
        print("Little higher")

    guess = int(input("Guess again: "))

print("You got it right!")
```
